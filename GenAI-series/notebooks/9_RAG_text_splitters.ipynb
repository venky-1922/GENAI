Text splitter

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(chunk_size=100, chunk_overlap=20)
chunks = text_splitter.split_text("This is a sample text to demonstrate the text splitting functionality. Langchain provides various text splitters to help you break down large texts into manageable chunks for processing.")
```

```python
print(chunks)
```
