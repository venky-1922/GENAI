Pydantic Data Validation

```python
from pydantic import BaseModel, Field

class User(BaseModel):
    name : str = Field(max_length=30)
    height: int = Field(lt=200, gt=0)
    age: int  = Field(lt=100,gt=0)
    summary: str = Field(min_length=20)


user = User (name="Venky",height=168,age=2,summary="Hi I am venky and I am learning GenAI currently")
user
```

```python
class Employee_details(BaseModel):
    Employee_id:int 
    Employee_details: User

Employee = Employee_details(Employee_id=123,Employee_details=user)
Employee
```

```python
from typing import List,Annotated

class Employees(BaseModel):
    users : List[User]
    comment : str = Field(
        default= "No comment"
    )
    
user2 = User (name="Vinay",height=168,age=2,summary="Hi I am venky and I am learning GenAI currently")
Employees_details = Employees(users=[user, user2],comment="users")
for user in Employees_details.users:
    print(user.name)

```

```python
Common_Validator = Annotated[List[User],"This is common validator",Field(default=[])]
class Company(BaseModel):
    Employees : Common_Validator
    id:str

```

```python
Company_details = Company(Employees=[user,user2],id="50c67abc")
Company_details
```
