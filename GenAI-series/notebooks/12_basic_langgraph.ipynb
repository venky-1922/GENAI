```python
from pydantic import BaseModel
from langgraph.graph import StateGraph, START, END
```

```python
class GreetState(BaseModel):
    message:str = ""

graph = StateGraph(GreetState)
```

```python
def Greet(state : GreetState) :
    state.message = f"{state.message} All Good !"
    return state

def UpperCase(state : GreetState):
    state.message = state.message.upper()
    return state

```

```python
graph.add_node("greet",Greet)
graph.add_node("upper case", UpperCase)
```

```python
graph.add_edge(START,"greet")
graph.add_edge("greet","upper case")
graph.add_edge("upper case", END)
```

```python
response = graph.compile()
result = response.invoke({"message":"I love you"})
result
```
