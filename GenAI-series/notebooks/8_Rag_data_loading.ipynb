Unstructured pdf Loader


```python
from pdf2image import convert_from_path
import pytesseract

images = convert_from_path("../data/au_exams.pdf", dpi=300)

text = ""
for img in images:
    text += pytesseract.image_to_string(img)

print(text)
```
